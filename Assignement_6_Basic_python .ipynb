```python
#1. Ramesh’s basic salary is input through the keyboard. His dearness allowance is 40% of basic salary, and
#house rent allowance is 20% of basic salary. Write a program to calculate his gross salary.
```

```python
basic_salary=eval(input('enter a basic salary :'))
dearness_allowance=(40*basic_salary)/100
rent_allowance=(20*basic_salary)/100
gross_salary=basic_salary+dearness_allowance+rent_allowance
print(f'gross salary of Ramesh is : {gross_salary}')
```

```python
#2. The distance between two cities (in km.) is input through the keyboard. Write a program to convert and
#print this distance in meters, feet, inches and centimeters.
distance_of_two_cities_in_km=eval(input('enter distance between two cities in km :'))
meters = distance_of_two_cities_in_km * 1000
centimeters = distance_of_two_cities_in_km * 100000
feet = distance_of_two_cities_in_km* 3280.84
inches = distance_of_two_cities_in_km* 39370.08
print(f'distance between two cities in meters :{meters}')
print(f'distance between two cities in centimeters :{centimeters}')
print(f'distance between two cities in feet :{feet}')
print(f'distance between two cities in inches :{inches}')
```

```python
#Temperature of a city in Fahrenheit degrees is input through the keyboard. Write a program to convert this
#temperature into Centigrade degrees.
fahrenheit=eval(input('enter a temperature of city in Fahrenheit degrees : '))
centigrade=5/9*(fahrenheit-32)
print(f'temperature of city in centigrade degress is {centigrade} when farenheit {fahrenheit}')
```
